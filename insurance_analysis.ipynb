## Data Cleaning

```python
#  (NaN) check
df.isnull().sum().sort_values(ascending=False).head(5)
```

```python
# "?" wali values har column mein check
(df == "?").sum()[lambda s: s > 0]
```

```python
# Duplicate rows
df.duplicated().sum()
```

```python
# _c39 htana 
df = df.drop(columns=["_c39"])
df.shape
```

```python
df["authorities_contacted"].isnull().sum()
```

```python
df["authorities_contacted"] = df["authorities_contacted"].fillna("None")
df.to_csv("insurance_claims_clean.csv", index=False)
```

## Fix Missing Value and New Column

```python
#  collision_type ka "?" kin incident types mein hai
pd.crosstab(df["incident_type"], df["collision_type"])
```

```python
# "?" ko samajhne name main replace
df["collision_type"] = df["collision_type"].replace("?", "No Collision")
df["property_damage"] = df["property_damage"].replace("?", "Unknown")
df["police_report_available"] = df["police_report_available"].replace("?", "Unknown")
```

```python
# Date columns ko text se date type mein badalna
df["policy_bind_date"] = pd.to_datetime(df["policy_bind_date"])
df["incident_date"] = pd.to_datetime(df["incident_date"])
df[["policy_bind_date", "incident_date"]].dtypes
```

```python
# Age group, incident ka din aur fraud ka Yes/No
df["age_group"] = pd.cut(df["age"], bins=[17, 30, 45, 60, 100], labels=["18-30", "31-45", "46-60", "60+"])
df["incident_day"] = df["incident_date"].dt.day_name()
df["fraud_flag"] = df["fraud_reported"].map({"Y": "Yes", "N": "No"})
df[["age_group", "incident_day", "fraud_flag"]].head()
```

```python
#  injury + property + vehicle claim = total claim?
(df["injury_claim"] + df["property_claim"] + df["vehicle_claim"] == df["total_claim_amount"]).mean()
```

```python
# Fraud check
df["fraud_flag"].value_counts()
```

```python
# Cleaned data save karna
df.to_csv("insurance_claims_clean.csv", index=False)
```

```python
# Dashboard ke numbers Python se dobara nikalna, Power BI se match karne ke liye
print("Total claims:", len(df))
print("Total amount (Lakh):", round(df["total_claim_amount"].sum() / 100000, 2))
print("Avg claim:", round(df["total_claim_amount"].mean(), 2))
fraud = df[df["fraud_flag"] == "Yes"]
print("Fraud claims:", len(fraud))
print("Fraud %:", round(len(fraud) / len(df) * 100, 1))
print("Fraud amount (Lakh):", round(fraud["total_claim_amount"].sum() / 100000, 2))
print("Avg fraud claim:", round(fraud["total_claim_amount"].mean()))
print(df["incident_type"].value_counts(), "\n")
print(df["incident_severity"].value_counts(), "\n")
print(df["age_group"].value_counts(), "\n")
print(df["incident_day"].value_counts(), "\n")
print((df.groupby("incident_state")["total_claim_amount"].sum() / 100000).round(2).sort_values(ascending=False), "\n")
print((df[["injury_claim", "property_claim", "vehicle_claim"]].sum() / df["total_claim_amount"].sum() * 100).round(2), "\n")
for col in ["incident_type", "incident_severity", "police_report_available", "age_group"]:
    print((df.groupby(col, observed=True)["fraud_flag"].apply(lambda s: (s == "Yes").mean() * 100)).round(1).sort_values(ascending=False), "\n")
```
